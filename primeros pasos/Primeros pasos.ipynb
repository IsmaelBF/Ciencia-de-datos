# Funciones

```python
def clasificar_edad (edad):
    """ Recibe una edad y regresa la categoria correspondiente"""
    if edad < 12:
        return "niño"
    elif edad < 18:
        return "adolescente"
    elif edad < 65:
        return "adulto"
    else:
        return "adulto mayor"

#Probemos la funcion con distintas edades

for edad in [5,15,40,80]:
    print(edad, "->", clasificar_edad(edad))
```
