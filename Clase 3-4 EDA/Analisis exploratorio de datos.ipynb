```python
import pandas as pd #analisis de datos
import numpy as np  #carga de datos
import matplotlib.pyplot as plt #graficas
```

```python
df = pd.read_csv('titanic.csv')
print("carga de datos exitosa")
```

# **Limpieza basica de datos**
#### vamos a tratar con valores faltantes, rellenar las comlumnas que tienen datos faltantes 
#### dropear cabin y embarked
#### convertir los tipos: asegurar  que la coñlumna plcase sea categorico

```python
df['Age'] = df['Age'].fillna(df['Age'].median()) #Rellenar columnas de Age
```

```python
df.head(10)
```

```python
df.dropna(subset=['Embarked'], inplace=True) #eliminar filas
```

```python
df.drop('Cabin', axis=1, inplace=True) #eliminar la columna)
```

```python
df['Pclass']=df['Pclass'].astype('category')#convierte en categoria
```

```python
df.info()
```

# **ANALISIS UNIVARIADO**

```python
import seaborn as sns
sns.set(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"]=(10,6)
```

```python
#variable objetivo
sns.countplot(x="Survived", data=df)
plt.title("Distribucion de sobrevivientes")
plt.show()
```

```python
cat_vars= ["Sex","Pclass","Embarked"]
```

```python
for col in cat_vars:
    sns.countplot(x=col,data=df)
    plt.title(f"Distribucion de col {col}")
    plt.show()
```

```python
num_vars=["Age","Fare"] # variables numericas
```

```python
for col in num_vars:
    sns.histplot(df[col].dropna(),kde=True, bins=30)
    plt.title(f"Distrubucion de {col}")
    plt.show()
```

# **Analisis bivariable (dos variables)**

```python
#Supervivencia por sexo
sns.barplot(x="Sex", y="Survived", data=df)
plt.title("Sobrevivientes según sexo")
plt.show()
```

```python
#Supervivencia por clase
sns.barplot(x="Pclass", y="Survived", data=df)
plt.title("Sobrevivientes según clase")
plt.show()
```

```python
#supervivecia por edad
sns.barplot(x="Survived", y="Age", data=df)
plt.title("Sobrevivientes segun edad")
plt.show()
```

# **COLUMNAS DE CORRELACIONES**

```python
#CORRELACIONES
corr=df.corr(numeric_only=True)
sns.heatmap(corr,annot=True, cmap= "coolwarm")
plt.title("Mapa de correlaciones")
plt.show()
```
