```python
# explicacion del uso de un for con tatblas de multiplicar
# cree una funcion que imprima la tabla a de multiplicar que el usuario solicite
def crear_tablas (numero):
    for i in range(11):
        print(f"{i} x {numero} = {numero*i}")
        
print("ingrese el numero del cual quiere su tabla de multiplicar")
numero = int(input())
crear_tablas(numero)
```
